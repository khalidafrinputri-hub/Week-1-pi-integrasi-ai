### 3. Siapkan konfigurasi postcell

Konfigurasi di bawah menggunakan ID kelas baru. Jangan mengubah `instructor_id` atau `class_id` kecuali instruktur meminta.

```python
import json
from pathlib import Path

postcell_config = {
    "url": "https://postcell.io/post_cell",
    "student_id": student_name.strip().replace(" ", "_"),
    "instructor_id": "AsYVdRr6HBPiKVHEvsTNhN13nBo1",
    "class_id": "perempuan_inovasi_ai",
    "should_send_to_server": "true",
}

# Simpan konfigurasi ke file lokal pada runtime Colab.
# Catatan: nama kelompok belum menjadi field di schema konfigurasi postcell.
with open("postcell.conf", "w", encoding="utf-8") as conf:
    json.dump(postcell_config, conf, indent=4)

print("Konfigurasi dibuat:")
print(json.dumps(postcell_config, indent=4))
print("Kelompok yang dipilih:", student_group.value)
```
